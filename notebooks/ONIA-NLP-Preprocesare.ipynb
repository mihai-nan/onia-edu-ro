# NLP — corpus, curățare și tokenizare
Parcurge celulele în ordine. Exemplu cu date sintetice în română; nu există dependențe de fișiere externe.

```python
import re
import pandas as pd
from io import StringIO
csv = "id,text,clasa\n1,Nu îmi place oferta,negativ\n2,Oferta este bună,pozitiv\n3,,necunoscut"
df = pd.read_csv(StringIO(csv))
df["text"] = df["text"].fillna("")
print(df.to_string(index=False))
```

## Statistici exploratorii
Calculăm lungimile, distribuția claselor și frecvențele termenilor.

```python
from collections import Counter
print(df["clasa"].value_counts())
df["nr_cuvinte"] = df["text"].str.split().str.len()
print(df[["id", "nr_cuvinte"]])
print(Counter(" ".join(df["text"]).casefold().split()).most_common(8))
```

## Transformări cu regex și Unicode
Păstrăm negația deoarece poate schimba sensul.

```python
import unicodedata
def curata(t):
    t = unicodedata.normalize("NFC", t)
    t = re.sub(r"https?://\S+", " URL ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t.casefold()
for t in df["text"]:
    print(curata(t))
```
