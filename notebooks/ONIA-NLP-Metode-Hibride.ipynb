# NLP — teme, combinare de caracteristici și pseudoetichetare

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.cluster import KMeans
docs=["tenis sport finala","fotbal sport meci","python cod date","python algoritmi model","sport echipa meci","date știință python"]
X=TfidfVectorizer().fit_transform(docs)
svd=TruncatedSVD(n_components=2,random_state=0)
Z=svd.fit_transform(X)
print(np.round(Z,3))
```

## Cluster ca informație complementară

```python
km=KMeans(n_clusters=2,random_state=0,n_init=10)
cl=km.fit_predict(Z)
print(cl)
print("Distanțe către centre",np.round(km.transform(Z),2))
```

## Alegerea pseudoetichetelor în funcție de încredere

```python
prob=np.array([.94,.81,.56,.22,.08])
for prag in [.7,.9]:
    confidence=np.maximum(prob,1-prob)
    selected=np.where(confidence>=prag)[0]
    print(prag,"alese",selected.tolist())
```
