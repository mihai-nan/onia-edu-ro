# NLP — similaritate lexicală, căutare și clustering

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
docs=["python cod programare", "date machine learning", "tenis meci sport", "fotbal competiție sport", "programare python date", "tenis finală meci"]
vec=TfidfVectorizer()
X=vec.fit_transform(docs)
q=vec.transform(["python și date"])
s=cosine_similarity(q,X)[0]
for i in s.argsort()[::-1][:3]: print(i,round(s[i],3),docs[i])
```

## K-Means și termenii reprezentativi

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=2,random_state=2,n_init=10)
y=model.fit_predict(X)
print(y)
words=vec.get_feature_names_out()
for i,c in enumerate(model.cluster_centers_):
    print("cluster",i,words[c.argsort()[-4:][::-1]].tolist())
```

## NMF — extragerea temelor

```python
from sklearn.decomposition import NMF
nmf=NMF(n_components=2,init="nndsvda",max_iter=500,random_state=1)
W=nmf.fit_transform(X)
print(W.shape,nmf.components_.shape)
for i,c in enumerate(nmf.components_):print(i, words[c.argsort()[-4:][::-1]].tolist())
```
